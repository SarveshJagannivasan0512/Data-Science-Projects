Incase of LLM's we have 'System message'(Manages the persona of LLM) and 'user message'(Manages the user prompt).

```python
# zero shot prompting using Gemini llm and langchain
# Use case: Product positioning for an AI based product.

zero_shot_prompt=ChatPromptTemplate.from_messages([
    ('system','You are a senior marketing strategist') ,# persona attached to the LLM
('human',"""Create a concise product positioning statement for an AI powered resume screening platform.
The target audience for this product are the HR-Leaders from mid-sized companies.
Ensure that the response generated is within 60 words and in a professional tone""") #user query
    ])
```

```python
response=llm.invoke(zero_shot_prompt.format_messages()) # the format message will pass the prompt according to LLM format(json) and understandibility
print(response)
```

```python
# Chain of thought prompting using Gemini LLM and Langchain for BFSI
# Objective: Loan eligibility assessment


prompt_2=ChatPromptTemplate.from_messages([
    ('system','You are a credit risk analyst at a commercial bank'),
    ('human',"""
    Assess the loan eligibility using the following structured evaluation method provided below as reference.

    Applicant Details:
    - Credit score: 800
    - Monthly Income: 100000
    - existing emi obligation: 12000 INR
    - Loan amount requested: 30,00,000 INR

    Evaluation Steps:
    - Credit worthiness assessment
    - Affordability analysis
    - Risk considerations
    - Final eligibility decision

    Output Format:
    - Clear section of headers
    - Bullet points clearly stated with reasons
    - Final decision clearly stated with reason"""
    )
])
```

```python
response=llm.invoke(prompt_2.format_messages())
print(response)
```

Chain of Thought Prompting in sequential manner
- Title
- Short Outline around the title
- Detailed Blog
- Summarize the detailed blog in 3-5 sentences


```python
llm3=ChatGoogleGenerativeAI(model='gemini-2.5-flash',temperature=0.8)
```

```python
from langchain_core.language_models import LLM
from langchain_core.prompts.string import PromptTemplateFormat
from langchain_core.prompts import PromptTemplate
from tempfile import template
# Chain1- Generating the blog title

title_prompt=PromptTemplate.from_template('Create one short catchy blog title about the {topic}')
title_chain=title_prompt | llm3
response=title_chain.invoke({'topic':'AI'})
print(response.content)
```

```python

prompt_3=ChatPromptTemplate.from_messages([
    ('system','You are a blog creater'),
    ('human',"""

    Create a blog with the following output format below as reference:

    Blog title:
    - How can the current Tamil Nadu Ranji trophy team get strenthened to win titles?

    - From the title create an outline
    - From the outline create a blog
    - From the blog create a 3-4 line summary

    Output Format:
    - Clear section of headers outline
    - Bullet points clearly stated with reasons
    - Final decision clearly stated with reason"""
    )
])
```

```python
responses3=llm3.invoke(prompt_3.format_messages())
print(responses3.content)
```
